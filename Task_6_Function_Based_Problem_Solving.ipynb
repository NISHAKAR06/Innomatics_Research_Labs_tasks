Problem 1: Smart Parking Lot Management System

```python
def smart_parking_system(capacity, logs):
    """
    Manages parking lot entries and exits.
    Returns current parked vehicles and parking status.
    """
    parked = 0
    peak = 0

    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1
        peak = max(peak, parked)
        if parked > capacity:
            print("⚠ Alert: Parking capacity exceeded!")

    status = "Available" if parked < capacity else "Full"
    print("Currently Parked Vehicles:", parked)
    print("Peak Usage:", peak)
    print("Parking Status:", status)
smart_parking_system(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

Problem 2: Online Food Delivery Time Estimator

```python
def estimate_delivery_time(distance, traffic, weather):
    """
    Estimates delivery time based on distance, traffic, and weather.
    """
    eta=distance*5

    if traffic=="High":
        eta += 15
    elif traffic=="Medium":
        eta += 8
    if weather=="Rainy":
        eta += 10
    elif weather=="Storm":
        eta+=20

    print("Estimated Delivery Time:", eta, "minutes")

estimate_delivery_time(8, "High", "Rainy")
```

Problem 3: Movie Theatre Seat Occupancy Analyzer

```python
def seat_occupancy_analyzer(total_seats, booked_seats):
    """
    Calculates occupancy percentage and show status.
    """

    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    if occupancy == 100:
        status = "Housefull"
    elif occupancy >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"

    print("Occupancy:", f"{occupancy:.0f}%")
    print("Show Status:", status)

seat_occupancy_analyzer(200, [1] * 150)
```

Problem 4: Cloud Server Load Classification System

```python
def server_load_classifier(cpu_readings):
    """
    Classifies server load based on average CPU usage.
    """

    average = sum(cpu_readings) / len(cpu_readings)

    if average < 50:
        status = "Normal"
    elif 50 <= average <= 80:
        status = "Warning"
    else:
        status = "Critical"

    print("Average CPU Load:", f"{average:.0f}%")
    print("Server Status:", status)

server_load_classifier([45, 60, 70, 85, 90])
```

Problem 5: Smart Classroom Resource Usage Monitor

```python
def classroom_resource_monitor(resource_usage):
    """
    Identifies overused classroom resources.
    """
    overused=[]

    for resource,hours in resource_usage.items():
        if hours>8:
            overused.append(resource)

    if overused:
        print("Overused Resources:",",".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")
classroom_resource_monitor({
    "Projector": 6,
    "AC": 9,
    "Lights": 4
})
```

Problem 6: Online Event Registration Capacity Controller

```python
def event_registration_controller(capacity, registrations):
    """
    Manages event registrations and prevents overbooking.
    """
    if registrations<=capacity:
        confirmed=registrations
        waitlisted=0
        status="Open"
    else:
        confirmed=capacity
        waitlisted=registrations - capacity
        status="Closed"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

event_registration_controller(100, 105)
```
