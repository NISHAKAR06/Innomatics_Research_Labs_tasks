Problem 1: Social Media – Post Engagement Analyzer

```python
def analyze_engagement(l):
    t=0
    for i in l:
        t+=i
    s="Viral Post" if t>=1000 else "Normal Engagement"
    print("Total Likes:",t)
    print("Post Status:",s)
analyze_engagement([200,150,300,400])
```

Problem 2: Healthcare – Medicine Stock Alert System

```python
def check_medicine_stock(s):
    st="Low Stock Alert" if s<10 else "Stock Sufficient"
    print("Medicine Stock:",s)
    print("Status:",st)
check_medicine_stock(6)
```

Problem 3: Agriculture – Rainfall Adequacy Checker

```python
def rainfall_checker(r,rq):
    t=0
    for i in r:
        t+=i
    a=t/len(r)
    s="Adequate Rainfall" if a>=rq else "Inadequate Rainfall"
    print("Average Rainfall:",a)
    print("Rainfall Status:",s)
rainfall_checker([70,75,68,80,67],70)
```

Problem 4: Social Media – Duplicate Account Detection

```python
def detect_duplicates(u):
    s="Yes" if len(set(u))!=len(u) else "No"
    print("Duplicate Accounts Found:",s)
detect_duplicates(["john","alice","john","bob"])
```

Problem 6: Agriculture – Premium Crop Price Filter

```python
def premium_crop_filter(p):
    pr=[]
    for i in p:
        if i>2000:
            pr.append(i)
    print("Premium Crops:",pr)
premium_crop_filter([1500,2500,1800,3200])
```

Problem 7: System Monitoring – Application Health Checker

```python
def application_health_checker(e):
    if e==0:
        s="Healthy"
    elif e<=5:
        s="Minor Issues"
    else:
        s="Critical Issues"
    print("Error Count:",e)
    print("System Status:",s)
application_health_checker(7)
```

Problem 9: E-Learning – Student Attendance Eligibility System

```python
def attendance_checker(a):
    t=len(a)
    p=0
    for i in a:
        if i=="Present":
            p+=1
    per=(p/t)*100
    s="Eligible" if per>=75 else "Not Eligible"
    print("Attendance Percentage:",per)
    print("Exam Eligibility:",s)
attendance_checker(["Present","Absent","Present","Present","Present"])
```

Problem 10: Smart Electricity Bill Analyzer

```python
def electricity_bill(u):
    if u<=100:
        b=u*3
    elif u<=200:
        b=100*3+(u-100)*5
    else:
        b=100*3+100*5+(u-200)*7
    if b<500:
        s="Low Usage"
    elif b<=1500:
        s="Moderate Usage"
    else:
        s="High Usage"
    print("Units Consumed:",u)
    print("Total Bill:",b)
    print("Usage Classification:",s)
electricity_bill(250)
```
